## Step - 4: Handling missing Values

### Objective:

Fill missing values using suitable statistical methods.

We will use:
    
- Median for Age.
- Mode for Gender.
- Mode for City.
- Median for Purchase Amount.
- Median for Feedback Score.
- "India" for missing Country, because all non-missing country values represent India.

```python
# Fill Age using median
df['Age'] = pd.to_numeric(
    df['Age'].astype(str).str.extract(r'(\d+(?:\.\d+)?)')[0],
    errors='coerce'
)

df.loc[~df['Age'].between(18, 100), 'Age'] = np.nan
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
# Fill Gender using mode
df['Gender'] = df['Gender'].astype(str).str.strip().str.lower()

df['Gender'] = df['Gender'].replace({
    'm': 'Male',
    'male': 'Male',
    'f': 'Female',
    'female': 'Female'
})

df['Gender'] = df['Gender'].replace('nan', np.nan)
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
```

```python
# Fill City using mode
df['City'] = df['City'].astype(str).str.strip().str.title()
df['City'] = df['City'].replace('Nan', np.nan)
df['City'] = df['City'].fillna(df['City'].mode()[0])
```

```python
# Fill Feedback Score
df['Feedback_Score'] = df['Feedback_Score'].fillna(
    df['Feedback_Score'].median()
)
```

### Explanation

Missing values in numerical columns were handled using median imputation, while categorical columns were handled using more imputation. This helps retain useful records without unnecessarily deleting large amount of data.

## Step- 6 : Correct inconsistent categorical values

## Explanation:

# Step- 8: Handle inavlid values and outliers

## objective: 

Identify and handle unrealistic values.

```python
# Handle invalid ages
df.loc[
    ~df['Age'].between(18, 100),
    'Age'
] = np.nan
```

```python
# Handle negative and unusually high purchase amounts
df.loc[
    (df['Purchase_Amount'] < 0) |
    (df['Purchase_Amount'] > 100000),
    'Purchase_Amount'
] = np.nan
```

```python
# Handle invalid purchase dates
df.loc[
    df['Last_Purchase_Date'] < df['Signup_Date'],
    'Last_Purchase_Date'
] = pd.NaT
```

After identifying invalid values, missing values created by these corrections should be imputed again:

```python
df['Age'] = df['Age'].fillna(df['Age'].median())

df['Purchase_Amount'] = df['Purchase_Amount'].fillna(
    df['Purchase_Amount'].median()
)
```

## Explanation:

Invalid values were identified and handled to improve dat accuracy. Negative purchase amounts and unusuallly high purchase amounts were treated as invalid as invalid for this analysis.Purchase dates occuring before customer signup dates were converted into missing values because they are chrnologically inconsistent.

## Step-9: Create derived columns

### Objective: 
Create new columns from existing data to support further analysis.
We will create:
- Signup Year
- Signup Month
- Age Group
- Days Between Signup and Purchase
- Days Since Last Purchase

```python
 #Extract signup year
df['Signup_Year'] = df['Signup_Date'].dt.year
```

```python
# Extract signup month
df['Signup_Month'] = df['Signup_Date'].dt.month
```

```python
# Create age groups
df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[17, 25, 35, 45, 55, 65, 100],
    labels=['18-25', '26-35', '36-45', '46-55', '56-65', '66+']
)
```

```python
# Calculate days between signup and purchase
df['Days_Between_Signup_And_Purchase'] = (
    df['Last_Purchase_Date'] - df['Signup_Date']
).dt.days
```

```python
# Calculate days since last purchase
reference_date = df['Last_Purchase_Date'].max()

df['Days_Since_Last_Purchase'] = (
    reference_date - df['Last_Purchase_Date']
).dt.days

df.head()
```

### Explanation:
New derived columns were created to provide additional information about customer behaviour. These columns will help analyse customer signup trends, age groups, purchasing patterns and customer activity
